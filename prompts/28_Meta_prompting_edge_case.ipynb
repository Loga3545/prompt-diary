# Промпт #28 — Meta-prompting edge case (плохой промпт)

**Техника:** Meta-prompting  
**Задача:** Улучшить намеренно плохой промпт через meta-prompting  
**Сложность:** ⭐⭐⭐⭐☆

```python
import sys
sys.path.append('..')
from config import get_completion
```

```python
# Намеренно плохой промпт — расплывчатый, без структуры
bad_prompt = "напиши что-нибудь про маркетинг"

# Шаг 1 — запускаем плохой промпт
print("=== РЕЗУЛЬТАТ С ПЛОХИМ ПРОМПТОМ ===")
response_bad = get_completion(bad_prompt)
print(response_bad)

# Шаг 2 — meta-промпт улучшает плохой промпт
meta_prompt = f"""Ты эксперт по prompt engineering.
Вот плохой промпт: "{bad_prompt}"

Он плохой потому что:
- Нет роли
- Нет контекста
- Нет формата вывода
- Нет ограничений

Улучши его. Верни только текст улучшенного промпта, без пояснений."""

improved_prompt = get_completion(meta_prompt)
print("\n=== УЛУЧШЕННЫЙ ПРОМПТ ===")
print(improved_prompt)

# Шаг 3 — запускаем улучшенный промпт
print("\n=== РЕЗУЛЬТАТ С УЛУЧШЕННЫМ ПРОМПТОМ ===")
response_improved = get_completion(improved_prompt)
print(response_improved)

# Сравнение длины
print("\n=== СРАВНЕНИЕ ===")
print(f"Плохой промпт → {len(response_bad.split())} слов")
print(f"Улучшенный промпт → {len(response_improved.split())} слов")
```

## Оценка: 5/5

## Инсайт
Meta-prompting полностью трансформировал расплывчатый запрос.

Плохой промпт "напиши что-нибудь про маркетинг" → общая Wikipedia-статья
без аудитории, без цели, без структуры. 243 слова ни о чём конкретном.

Улучшенный промпт добавил всё что отсутствовало:
- Роль: директор по маркетингу средней компании
- Объём: ~500 слов
- Тема: последние тенденции цифрового маркетинга
- Аудитория: 25-45 лет
- Ограничение: без отраслевой терминологии

Результат: структурированная статья с заголовками, примерами
кампаний и конкретными советами. 357 слов по делу.

Главный вывод: meta-prompting особенно ценен когда знаешь ЧТО
хочешь но не знаешь КАК это правильно сформулировать.
Модель лучше знает что делает хороший промпт — дай ей это сделать.
Два запроса вместо одного, но качество несравнимо выше.
