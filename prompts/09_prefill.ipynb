# Промпт #09 — Prefill

**Техника:** Prefill (предзаполнение ответа)  
**Задача:** Управлять форматом вывода через prefill  
**Сложность:** ⭐⭐☆☆☆

```python
import sys
sys.path.append('..')
from config import get_completion
```

```python
review = "Ноутбук быстрый, но батарея слабая и цена высокая."

# Без prefill
prompt = f"""Проанализируй отзыв и выдели плюсы и минусы.

Отзыв: {review}"""

print("Без prefill:")
print(get_completion(prompt))
print("\n" + "="*50 + "\n")

# С prefill — заставляем начать с JSON
print("С prefill:")
print(get_completion(prompt, prefill='{"плюсы":'))
```

## Оценка: 5/5

## Инсайт
Prefill принудительно задаёт формат — модель продолжает 
с того места где мы начали.

**Сравнение:**
- Без prefill: длинный текст с markdown заголовками
- С prefill `{"плюсы":`: чистый JSON без лишнего текста ✅

**Применение:**
- Нужен JSON → начни с `{`
- Нужен список → начни с `1.`
- Нужен XML → начни с `<result>`

**Вывод:** prefill — самый простой способ гарантировать 
нужный формат без длинных инструкций.
