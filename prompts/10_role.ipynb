**Задача:** Сравнить ответ с ролью и без роли

```python
import sys
sys.path.append('..')
from config import get_completion
```

```python
question = "Как мне начать инвестировать с нуля?"

# Без роли
print("Без роли:")
print(get_completion(question))
print("\n" + "="*50 + "\n")

# С ролью
print("С ролью:")
print(get_completion(
    question,
    system_prompt="Ты опытный финансовый советник с 20-летним стажем. Отвечаешь чётко и практично, без воды."
))
```

```python
question = "Как мне начать инвестировать с нуля?"

print("Без роли:")
print(get_completion(question))
```

```python
print("С ролью:")
print(get_completion(
    question,
    system_prompt="Ты опытный финансовый советник с 20-летним стажем. Отвечаешь чётко и практично, без воды."
))
```

## Оценка: 5/5

## Инсайт
Роль кардинально меняет качество и стиль ответа.

**Сравнение:**
- Без роли: 10 пунктов, много воды, американские брокеры ❌
- С ролью: 8 чётких шагов, российские брокеры, без лишнего текста ✅

**Ключевые отличия:**
- Роль адаптировала ответ под локальный контекст (рубли, ВТБ, Сбербанк)
- Ответ стал короче и практичнее
- Исчезли общие фразы типа "это отличный первый шаг"

**Вывод:** роль в system prompt — это не просто стиль,
она меняет содержание ответа. Модель думает от лица эксперта
и даёт более релевантные советы.
