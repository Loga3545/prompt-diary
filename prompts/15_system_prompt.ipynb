# Промпт #15 — System prompt vs user prompt

**Техника:** System prompt  
**Задача:** Сравнить ответ без system prompt и с ним на одном запросе  
**Сложность:** ⭐⭐☆☆☆

```python
import sys
sys.path.append('..')
from config import get_completion
```

```python
user_prompt = "Объясни что такое инфляция"

# Вариант А — без system prompt
response_a = get_completion(user_prompt)
print("=== БЕЗ SYSTEM PROMPT ===")
print(response_a)

# Вариант Б — с system prompt
system = "Ты объясняешь сложные темы простым языком для детей 10 лет. Используй короткие предложения и простые аналогии."

response_b = get_completion(user_prompt, system_prompt=system)
print("\n=== С SYSTEM PROMPT ===")
print(response_b)
```

## Оценка: 5/5

## Инсайт
System prompt кардинально меняет стиль ответа на тот же вопрос.
Без него — академический текст с болдом, списками, терминами (монетарная политика, 
фискальная политика). С ним — аналогия с конфетами в 3 абзаца, понятная ребёнку.

Ключевое наблюдение: user prompt задаёт ЧТО спросить, system prompt задаёт 
КАК отвечать. Это разные уровни управления моделью.

Побочный эффект: без system prompt модель выдала исчерпывающий ответ (все причины, 
плюсы, минусы, инструменты). С system prompt — только суть. 
Иногда "меньше" = лучше.
