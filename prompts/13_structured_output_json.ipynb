# Промпт #13 — Structured output JSON

**Техника:** Структурированный вывод в JSON  
**Задача:** Извлечь данные из текста и вернуть в JSON формате  
**Сложность:** ⭐⭐⭐☆☆

```python
import sys
import json
sys.path.append('..')
from config import get_completion
```

```python
text = """Вакансия: Промпт-инженер в компании TechCorp.
Зарплата: 150 000 рублей в месяц.
Опыт: от 2 лет. Локация: Москва или удалённо.
Требования: знание Python, LangChain, опыт с LLM."""

prompt = f"""Извлеки данные из вакансии и верни строго в JSON формате.
Без дополнительного текста, только JSON.

<vacancy>{text}</vacancy>

Формат:
{{
    "должность": "",
    "компания": "",
    "зарплата": "",
    "опыт": "",
    "локация": "",
    "требования": []
}}"""

result = get_completion(prompt)
print(result)

# Проверяем что это валидный JSON
try:
    parsed = json.loads(result)
    print("\n✅ Валидный JSON!")
    print(parsed)
except:
    print("\n❌ Невалидный JSON")
```

## Оценка: 5/5

## Инсайт
Комбинация XML тегов для входных данных + JSON шаблон 
для вывода даёт чистый парсинг без лишнего текста.

**Ключевые приёмы:**
- `<vacancy>` изолирует входные данные от инструкций
- Шаблон JSON в промпте показывает точную структуру
- "Без дополнительного текста" — убирает преамбулу
- `json.loads()` проверяет валидность программно

**Применение в реальных задачах:**
- Парсинг вакансий
- Извлечение данных из резюме
- Структурирование новостей
- Парсинг отзывов в БД

**Вывод:** JSON + XML теги = надёжный способ 
получить структурированные данные готовые 
к записи в базу данных.
