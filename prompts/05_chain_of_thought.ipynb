```python
import sys
sys.path.append('..')
from config import get_completion
```

```python
# Задача-ловушка где интуитивный ответ неверный
question3 = """Бита и мяч вместе стоят 1100 рублей. 
Бита стоит на 1000 рублей дороже мяча. 
Сколько стоит мяч?"""

prompt_no_cot = f"""Ответь на вопрос одним числом.

Вопрос: {question3}"""

prompt_cot = f"""Ответь на вопрос, рассуждая пошагово.

Вопрос: {question3}"""

print("Без CoT:")
print(get_completion(prompt_no_cot))
print("\nС CoT:")
print(get_completion(prompt_cot))
```

## Оценка: 5/5

## Инсайт
CoT критически важен для задач-ловушек где интуитивный ответ неверный.

**Сравнение:**
- Без CoT: 100 рублей ❌ (интуитивный неверный ответ)
- С CoT: 50 рублей ✅ (пошаговое рассуждение даёт верный ответ)

**Вывод:** на простых задачах CoT не нужен — модель справляется сама.
Но на задачах где есть ловушка — CoT спасает от неверного ответа.
Всегда используй CoT когда задача требует многошагового рассуждения.
