# Промпт #02 — Zero-shot edge case

**Техника:** Zero-shot  
**Задача:** Проверить где zero-shot даёт сбой - неоднозначные отзывы  
**Сложность:** ⭐⭐☆☆☆

```python
import sys
sys.path.append('..')
from config import get_completion
```

```python
# Неоднозначный отзыв - есть и плюсы и минусы
review = "Ноутбук быстрый, но батарея садится за 2 часа и цена завышена."

prompt = f"""Классифицируй отзыв как ПОЗИТИВНЫЙ или НЕГАТИВНЫЙ.
Отвечай одним словом.

Отзыв: {review}"""

result = get_completion(prompt)
print(result)
```

## Оценка: 3/5

## Инсайт
Zero-shot плохо справляется с неоднозначными отзывами - 
модель вынуждена выбрать одно из двух даже когда отзыв смешанный.

**Решение:** добавить третий вариант НЕЙТРАЛЬНЫЙ в инструкцию,
или использовать few-shot с примером смешанного отзыва.
Это будет проверено в промптах #03-04.
