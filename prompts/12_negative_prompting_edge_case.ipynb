## Инсайт
Модель обходит ограничения через синонимы — 
запретить все слова невозможно.

**Что сделала модель:**
- "компьютер" → "системы"
- "данные" → "информация"
- "обучение" → "экспертиза и опыт"
- "алгоритм" → "закономерности"

**Вывод:** негативные инструкции работают лучше 
когда запрещаешь 2-3 конкретных слова или стиля.
Слишком много ограничений → модель просто 
находит синонимы и обходит их.

**Правило:** негативные инструкции для направления,
не для полного контроля.
