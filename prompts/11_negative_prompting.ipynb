# Промпт #11 — Negative prompting

**Техника:** Negative prompting (что модель НЕ должна делать)  
**Задача:** Сравнить ответ с ограничениями и без  
**Сложность:** ⭐⭐☆☆☆

```python
import sys
sys.path.append('..')
from config import get_completion
```

```python
question = "Объясни что такое блокчейн"

# Без ограничений
print("Без ограничений:")
print(get_completion(question))
print("\n" + "="*50 + "\n")

# С негативными инструкциями
print("С негативными инструкциями:")
print(get_completion(f"""Объясни что такое блокчейн.

Не используй:
- технические термины
- аналогии с цепочками
- слово "децентрализованный"

Объясни как будто говоришь с человеком который 
никогда не слышал об этом."""))
```

## Оценка: 5/5

## Инсайт
Негативные инструкции упрощают и сокращают ответ.

**Сравнение:**
- Без ограничений: технический текст с терминами, 
  слово "децентрализованный" встречается 3 раза ❌
- С негативными инструкциями: простое объяснение 
  через аналогию с деревней ✅

**Вывод:** негативные инструкции ("не используй X") 
работают точнее чем позитивные ("объясни просто") — 
модель точно знает чего избегать, а не угадывает 
что значит "просто".
