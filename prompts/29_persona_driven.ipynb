# Промпт #29 — Persona-driven prompting

**Техника:** Persona-driven prompting  
**Задача:** Сравнить role prompting и полную персону на одном вопросе  
**Сложность:** ⭐⭐⭐☆☆

```python
import sys
sys.path.append('..')
from config import get_completion
```

```python
question = "Стоит ли мне вкладывать деньги в криптовалюту?"

# Вариант А — простой role prompting
response_role = get_completion(
    question,
    system_prompt="Ты финансовый советник."
)
print("=== ROLE PROMPTING ===")
print(response_role)

# Вариант Б — полная персона
persona = """Ты Михаил, финансовый советник с 20 летним опытом работы 
в российских банках. Говоришь прямо и кратко, без воды. 
Скептически относишься к высокорисковым инвестициям.
Всегда приводишь конкретные цифры и факты.
Если видишь риск — говоришь об этом честно и прямо.
Никогда не даёшь советов которые могут навредить клиенту."""

response_persona = get_completion(
    question,
    system_prompt=persona
)
print("\n=== PERSONA-DRIVEN ===")
print(response_persona)
```
