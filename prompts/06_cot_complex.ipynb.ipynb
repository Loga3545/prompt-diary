# Промпт #06 — Chain-of-Thought сложная задача

**Техника:** Chain-of-Thought (CoT)  
**Задача:** Проверить CoT на задаче с несколькими условиями  
**Сложность:** ⭐⭐⭐⭐☆

```python
import sys
sys.path.append('..')
from config import get_completion
```

```python
question = """В классе 30 учеников. 
40% из них занимаются спортом. 
Из тех кто занимается спортом — половина также занимается музыкой. 
Сколько учеников занимаются и спортом и музыкой?"""

prompt_no_cot = f"""Ответь на вопрос одним числом.

Вопрос: {question}"""

prompt_cot = f"""Ответь на вопрос, рассуждая пошагово.

Вопрос: {question}"""

print("Без CoT:")
print(get_completion(prompt_no_cot))
print("\nС CoT:")
print(get_completion(prompt_cot))
```

## Оценка: 5/5

## Инсайт
Без CoT модель остановилась на промежуточном результате — 
приняла частичный ответ за финальный.

**Сравнение:**
- Без CoT: 12 ❌ (остановился на первом шаге)
- С CoT: 6 ✅ (прошёл все шаги до конца)

**Вывод:** CoT особенно важен когда задача требует 
нескольких последовательных вычислений — 
без него модель "срезает путь" и берёт промежуточное число.
