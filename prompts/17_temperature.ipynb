# Промпт #17 — Температура (creative vs precise)

**Техника:** Temperature  
**Задача:** Сравнить ответы при низкой (0.1) и высокой (0.9) температуре  
**Сложность:** ⭐⭐☆☆☆

```python
import sys
sys.path.append('..')
from config import get_completion
```

```python
prompt = "Придумай необычное название для кофейни"

response_low = get_completion(prompt, temperature=0.1)
print("=== ТЕМПЕРАТУРА 0.1 (точная) ===")
print(response_low)

response_high = get_completion(prompt, temperature=0.9)
print("\n=== ТЕМПЕРАТУРА 0.9 (креативная) ===")
print(response_high)
```

## Оценка: 3/5

## Инсайт
Разница есть, но неожиданная. При температуре 0.1 модель дала структурированный 
список из 8 названий с объяснением каждого. При 0.9 — более хаотичный вывод: 
сначала 2 варианта, потом ещё 8, потом ещё 5 "поэтичных".

Ожидалось: 0.1 = скучные предсказуемые названия, 0.9 = безумно креативные.
По факту: 0.1 = организованный ответ, 0.9 = менее организованный, но не сильно 
креативнее. "Кофейный квантовый скачок" выбивается, но в целом оба ответа похожи.

Вывод: температура сильнее влияет на СТРУКТУРУ и ПРЕДСКАЗУЕМОСТЬ ответа, 
чем на креативность. Для задач где нужен один конкретный ответ (классификация, 
JSON, факты) — низкая температура. Для генерации идей — высокая, но разница 
меньше чем кажется.
