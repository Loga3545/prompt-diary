# Промпт #23 — Tree-of-Thought

**Техника:** Tree-of-Thought (ToT)  
**Задача:** Выбор стратегии запуска продукта через несколько веток рассуждений  
**Сложность:** ⭐⭐⭐☆☆

```python
import sys
sys.path.append('..')
from config import get_completion
```

```python
# Без ToT — обычный промпт
prompt_no_tot = """Как лучше всего запустить новое мобильное приложение на рынок?"""

response_no_tot = get_completion(prompt_no_tot)
print("=== БЕЗ ToT ===")
print(response_no_tot)

# С ToT — три ветки рассуждений
prompt_tot = """Ты стратег по запуску продуктов. 

Задача: запустить новое мобильное приложение на рынок.

Предложи 3 разные стратегии запуска. Для каждой:
1. Название стратегии
2. Основные шаги
3. Плюсы и минусы
4. Оценка от 1 до 10

После анализа всех трёх — выбери лучшую и объясни почему."""

response_tot = get_completion(prompt_tot)
print("\n=== С ToT ===")
print(response_tot)
```

## Оценка: 5/5

## Инсайт
Разница между обычным промптом и ToT наглядная.

Без ToT — модель дала один универсальный план на 4 раздела. Полезно, 
но это просто "стандартный ответ" без анализа альтернатив.

С ToT — модель предложила 3 принципиально разные стратегии, оценила каждую
и выбрала победителя с аргументами. Партнёрство (8/10) победило взрывной 
запуск (7/10) и мягкий запуск (6/10).

Ключевое наблюдение: ToT заставляет модель думать вширь перед выбором.
Без явного требования она всегда идёт по пути наименьшего сопротивления —
даёт первый очевидный ответ. ToT вынуждает рассмотреть альтернативы.

Практический вывод: используй ToT когда задача имеет несколько 
равнозначных решений и нужно выбрать лучшее с обоснованием.
Для простых задач — избыточно.
