# Промпт #16 — System prompt edge case

**Техника:** System prompt vs user prompt (конфликт)  
**Задача:** Что победит — роль из system prompt или запрос пользователя?  
**Сложность:** ⭐⭐⭐☆☆

```python
import sys
sys.path.append('..')
from config import get_completion
```

```python
system = "Ты краткий ассистент. Отвечай максимум 2 предложениями. Никогда не превышай этот лимит."

user_prompt = "Объясни подробно квантовую физику: что это такое, как работает, какие основные принципы, где применяется и почему это важно для науки."

response = get_completion(user_prompt, system_prompt=system)
print("=== РЕЗУЛЬТАТ ===")
print(response)
print(f"\n=== ДЛИНА ОТВЕТА: {len(response.split())} слов ===")
```

## Оценка: 5/5

## Инсайт
System prompt победил. User просил подробное объяснение с принципами, 
применением и историей — модель уложилась в 2 предложения (56 слов).

Ключевое наблюдение: system prompt — это жёсткие рамки, user prompt работает 
внутри этих рамок. Пользователь не может "перебить" роль, заданную разработчиком.

Практический вывод: если строишь продукт и хочешь контролировать поведение 
модели — логику и ограничения ставь в system prompt, а не доверяй user prompt.
Это защита от непредсказуемых запросов пользователей.
