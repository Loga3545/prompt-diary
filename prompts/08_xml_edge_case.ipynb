**Задача:** Проверить что модель делает когда данных не хватает  
**Сложность:** ⭐⭐⭐☆☆

```python
import sys
sys.path.append('..')
from config import get_completion
```

```python
# Текст где намеренно нет некоторых данных
text = """Меня зовут Саша, увлекаюсь фотографией и готовкой."""

prompt = f"""Извлеки данные из текста и верни в XML формате.

<text>{text}</text>

Верни ответ строго в таком формате:
<person>
    <name></name>
    <age></age>
    <city></city>
    <job></job>
    <hobbies></hobbies>
</person>"""

result = get_completion(prompt)
print(result)
```

## Оценка: 5/5

## Инсайт
Модель не галлюцинирует — когда данных нет, оставляет теги пустыми.

**Сравнение с #07:**
- #07: все поля заполнены ✅
- #08: нет данных → пустые теги, не выдумывает ✅

**Вывод:** XML шаблон не только структурирует вывод, 
но и защищает от галлюцинаций — модель не заполняет 
поля выдуманными данными когда информации нет.

**Применение:** безопасно использовать для парсинга 
реальных текстов где данные могут отсутствовать.
