# Промпт #27 — Meta-prompting

**Техника:** Meta-prompting  
**Задача:** Модель пишет промпт → промпт запускается → получаем результат  
**Сложность:** ⭐⭐⭐⭐☆

```python
import sys
sys.path.append('..')
from config import get_completion
```

```python
task = "Написать короткое описание вакансии Python-разработчика для стартапа"

# Шаг 1 — обычный промпт без meta
print("=== БЕЗ META-PROMPTING ===")
response_basic = get_completion(task)
print(response_basic)

# Шаг 2 — meta-промпт генерирует промпт
meta_prompt = f"""Ты эксперт по prompt engineering.
Напиши идеальный промпт для следующей задачи:

Задача: {task}

Промпт должен содержать:
- Роль модели
- Контекст задачи
- Чёткие требования к формату вывода
- Ограничения

Верни только текст промпта, без пояснений."""

generated_prompt = get_completion(meta_prompt)
print("\n=== СГЕНЕРИРОВАННЫЙ ПРОМПТ ===")
print(generated_prompt)

# Шаг 3 — запускаем сгенерированный промпт
print("\n=== РЕЗУЛЬТАТ С META-ПРОМПТОМ ===")
response_meta = get_completion(generated_prompt)
print(response_meta)
```

## Оценка: 5/5

## Инсайт
Meta-prompting сработал — результат стал короче и структурированнее.

Без meta: длинный текст с болдом, списками, секцией "Если вы:",
контактами, общими фразами типа "динамично развивающийся стартап" — 
типичный шаблонный HR-текст.

С meta: чистые два раздела "Обязанности" и "Требования", 
не более 200 слов, без лишнего. Модель сама добавила ограничение
на длину и формат которых не было в исходном запросе.

Ключевое наблюдение: сгенерированный промпт добавил то что
мы забыли указать — роль рекрутера, лимит 200 слов, запрет
на контактную информацию. Модель знает что делает хороший промпт
лучше чем мы сформулировали задачу.

Практический вывод: meta-prompting полезен когда не знаешь
как правильно сформулировать задачу. Вместо того чтобы
итеративно улучшать промпт вручную — просишь модель
сделать это за тебя.
