# Промпт #24 — Tree-of-Thought edge case (избыточность)

**Техника:** Tree-of-Thought (ToT)  
**Задача:** Показать когда ToT избыточен и CoT справляется лучше  
**Сложность:** ⭐⭐⭐☆☆

```python
import sys
sys.path.append('..')
from config import get_completion
```

```python
# Простая математическая задача

# Вариант А — CoT
prompt_cot = """Реши задачу пошагово:
В магазине 3 полки. На каждой полке 8 книг. 
Продали 5 книг с первой полки и 3 книги со второй. 
Сколько книг осталось?"""

response_cot = get_completion(prompt_cot)
print("=== CoT ===")
print(response_cot)

# Вариант Б — ToT на той же задаче
prompt_tot = """Реши задачу тремя разными способами, оцени каждый, выбери лучший:
В магазине 3 полки. На каждой полке 8 книг. 
Продали 5 книг с первой полки и 3 книги со второй. 
Сколько книг осталось?"""

response_tot = get_completion(prompt_tot)
print("\n=== ToT ===")
print(response_tot)

# Сравнение длины
print("\n=== СРАВНЕНИЕ ===")
print(f"CoT: {len(response_cot.split())} слов")
print(f"ToT: {len(response_tot.split())} слов")
```

## Оценка: 5/5

## Инсайт
ToT дал в 3 раза больше слов (339 vs 108) но ответ одинаковый — 16 книг.
Все три ветки привели к одному результату разными путями.

Это наглядно показывает когда ToT избыточен: задача с однозначным 
математическим ответом не выигрывает от нескольких веток рассуждений.
CoT дал чистый пошаговый ответ за 108 слов — быстрее, дешевле, понятнее.

Правило выбора техники:
- Однозначный ответ (математика, факты, извлечение данных) → CoT
- Несколько равнозначных решений (стратегия, планирование, выбор) → ToT

Практический вывод: ToT стоит токенов и времени. Используй его только
когда реально нужно сравнить альтернативы. На простых задачах это
просто лишний шум вокруг правильного ответа.
