# Промпт #30 — Persona-driven edge case (выход за рамки экспертизы)

**Техника:** Persona-driven prompting  
**Задача:** Как персона реагирует на вопрос вне своей экспертизы?  
**Сложность:** ⭐⭐⭐☆☆

```python
import sys
sys.path.append('..')
from config import get_completion
```

```python
persona = """Ты Михаил, финансовый советник с 20 летним опытом работы 
в российских банках. Говоришь прямо и кратко, без воды. 
Скептически относишься к высокорисковым инвестициям.
Всегда приводишь конкретные цифры и факты.
Если видишь риск — говоришь об этом честно и прямо.
Если вопрос выходит за твою экспертизу — честно говоришь об этом
и направляешь к нужному специалисту."""

# Вопрос в рамках экспертизы
question_in = "Стоит ли сейчас покупать облигации?"
response_in = get_completion(question_in, system_prompt=persona)
print("=== ВОПРОС В РАМКАХ ЭКСПЕРТИЗЫ ===")
print(response_in)

# Вопрос вне экспертизы
question_out = "У меня болит спина уже неделю, что посоветуешь?"
response_out = get_completion(question_out, system_prompt=persona)
print("\n=== ВОПРОС ВНЕ ЭКСПЕРТИЗЫ ===")
print(response_out)

# Пограничный вопрос — смежная тема
question_border = "Как стресс от потери денег влияет на здоровье?"
response_border = get_completion(question_border, system_prompt=persona)
print("\n=== ПОГРАНИЧНЫЙ ВОПРОС ===")
print(response_border)
```

## Оценка: 5/5

## Инсайт
Три разных ситуации — три разных поведения персоны.

В рамках экспертизы (облигации): конкретные цифры, позиция,
рекомендация — персона работает как в #29.

Вне экспертизы (спина): персона честно отказалась и направила
к врачу. Ограничение "если вопрос вне экспертизы — скажи об этом"
сработало точно. Это важно для продуктов — чат-бот не должен
давать медицинские советы.

Пограничный вопрос (стресс и здоровье): персона не отказалась,
но ответила через свою призму — финансовый стресс, данные ВОЗ,
и в конце вернулась к своей теме "поэтому я против рискованных
инвестиций". Модель нашла мост между медициной и финансами.

Главный вывод: хорошая персона не просто меняет стиль —
она задаёт границы компетентности. Это критично для продуктов
где важно чтобы ассистент не выходил за свою роль.
Ограничения в персоне работают так же как negative prompting.
